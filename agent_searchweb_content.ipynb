```python
from crewai import LLM, Agent
from langchain_community.tools import DuckDuckGoSearchRun
from crewai.tools import tool

print(" OK")
```

```python
search_tool_instance = DuckDuckGoSearchRun()
```

```python
@tool("DuckDuckGo Search")
def search_tool(query: str) -> str:
    """A tool for searching the web using DuckDuckGo."""
    return search_tool_instance.run(query)
```

```python
import os
os.environ["GROQ_API_KEY"] = "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"  # ta clé Groq

from crewai import LLM
AGENT_LLM = LLM(model="groq/llama-3.1-8b-instant")
```

```python
from crewai import Agent
research_agent = Agent(
  role="Senior Research Analyst",
  goal="Analyze the content of the provided website and extract key insights.",
  backstory="You are a world-class research analyst known for concise and insightful summaries.",
  verbose=False,
  llm=AGENT_LLM,
  tools=[search_tool]
)
print(" Agent prêt avec Groq !")
```

```python
from crewai import Task

research_task = Task(
  description="Identify the main conclusions and insights from the given content, focusing on key data points and trends.",
  expected_output="A concise summary of the key findings, including any important numerical data or overarching themes.",
  agent=research_agent,
  tools=[search_tool]
)

print(" Research Task Created!")
```

```python
from crewai import Crew

crew = Crew(
  agents=[research_agent],
  tasks=[research_task],
  verbose=True # Changed from 2 to True for proper boolean value
)

print(" Crew Created!")
```

```python
# Cellule 4 - Lance le test
url = "https://www.legifrance.gouv.fr/jorf/id/JORFTEXT000049449852" # Replace with the actual URL you want to analyze
result = crew.kickoff(inputs={'url': url})
print(result)
```
